# Spring stroke grammar v0.3

Three isolated comparisons on the locked F2-C scene: continuous D skeletons, vector-aware mist transfer and depth-specific vegetation. The camera, projection, source DEM/NDVI, 70/30 foreground and scalar whitespace fields are unchanged.

Install from the repository root with `python -m pip install -e ".[research,dev]"`. Set `DATASCAPE_DATA_DIR` to the local folder containing the original TIFFs. Raw inputs are never written. Teal vegetation is a development overlay, not a palette change.

```python
from pathlib import Path
import os, json
from IPython.display import display, Image
from datascape.rendering.grammar_study import run
from datascape.strokes.vector import read_vectors
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "configs/scenes/seoul_spring_stroke_grammar_v03.json").is_file())
DATA = Path(os.environ.get("DATASCAPE_DATA_DIR", ROOT / "data/raw"))
OUT = Path(os.environ.get("DATASCAPE_GRAMMAR_OUTPUT", ROOT / "outputs/spring_stroke_grammar_v03"))
CONFIG = ROOT / "configs/scenes/seoul_spring_stroke_grammar_v03.json"
master = DATA / "SeoulNorth_v02_MASTER_30m.tif"
veg = DATA / "SeoulNorth_v02_VEG_10m.tif"
assert master.is_file() and veg.is_file(), "Set DATASCAPE_DATA_DIR to the original TIFF directory."
```

```python
record = run(master, veg, CONFIG, OUT, ROOT)
```

## D: continuity before strength

The same old A/B/C/V paths remain fixed. Raw ridge, occlusion and rock-edge supports grow around previously selected D locations. Endpoint bridges must pass tangent, source-depth, source-distance and hard-omission checks. Short parents are removed; spacing and a length budget keep D sparse. The final envelope grows from a weak entry to a strong center and releases early. No bridge crosses a zero core.

```python
display(Image(filename=str(OUT / "D_comparison.png")))
display(Image(filename=str(OUT / "D_details.png")))
record["D_metrics"]
```

## Mist transfer: identical unattenuated A proposals

B/C/D/V are hidden in this comparison. Binary uses only the fixed hard cores. Density uses seeded path selection against the unchanged continuous retention. Combined also shortens retained windows and stores per-vertex width, opacity and dimensionless intensity. Width/intensity are preview controls, not calibrated machine pressure. The existing fields supply bottom/near separation, middle valley breathing and distant-foot omission.

```python
display(Image(filename=str(OUT / "M_comparison.png")))
display(Image(filename=str(OUT / "mist_depth_details.png")))
record["mist_metrics"]
```

## Vegetation: fixed roots, distinct depth grammars

Terrain A/B/C/D stays fixed. Near trees have curved trunks, staggered branches, twigs and open foliage gestures; middle vegetation uses irregular clusters; far vegetation uses sparse ridge-supported touches. Three duplicated small hero-parent glyphs are consolidated at the same root sites. These are data-supported expressive symbols, not observed trees or species. Object-depth visibility lets front branches appear against deeper mist.

```python
display(Image(filename=str(OUT / "V_comparison.png")))
display(Image(filename=str(OUT / "V_details.png")))
record["vegetation_metrics"]
```

## Proposed combination

Combined A mist transfer + unchanged B/C + enveloped D + redesigned vegetation. The proposed JSON, development/monochrome SVG and physical two-panel preview all use the same stored vertices and profiles. The 40 mm gap is inserted only for physical presentation. Review at full size before selecting this as a production baseline; no G-code is generated.

```python
display(Image(filename=str(OUT / "proposed_physical.png")))
display(Image(filename=str(OUT / "proposed_monochrome.png")))
record["invariants"]
```
