# Spring Pair Ground Composition Refinement

F1 is raw-meter physical perspective from a low-DEM eye. F2 uses the same contiguous source scene with recorded depth-banded width/relief and upward unfolding. Previous P2/P3 remain unchanged. No paths, trees, buildings, waterbody identities, vegetation, color or G-code are generated.

Install `python -m pip install -e ".[research,dev]"`, select that environment in VSCode and set `DATASCAPE_DATA_DIR` to the original MASTER folder. The 600 x 900 mm content and 40 mm inserted physical gap remain unchanged. The 500 x 500 mm test support does not change rendering proportions.

```python
import os
from pathlib import Path
from IPython.display import Image, display
from datascape.rendering.ground_study import run

ROOT = next((p for p in [Path.cwd(), Path.cwd().parent] if (p / "pyproject.toml").is_file()), None)
if ROOT is None:
    raise FileNotFoundError("Run from the repository or notebooks directory.")
DATA_DIR = Path(os.environ.get("DATASCAPE_DATA_DIR", str(ROOT / "data/raw")))
MASTER = DATA_DIR / "SeoulNorth_v02_MASTER_30m.tif"
CONFIG = ROOT / "configs/scenes/seoul_ground_composition_v01.json"
PREVIOUS = ROOT / "docs/experiments/spring_production_v01"
OUTPUT = Path(os.environ.get("DATASCAPE_GROUND_OUTPUT", str(ROOT / "outputs/spring_ground_composition_v01")))
if not MASTER.is_file():
    raise FileNotFoundError("Set DATASCAPE_DATA_DIR to the original MASTER folder.")
```

```python
record = run(MASTER, CONFIG, OUTPUT, PREVIOUS)
assert record["source_preserved"] and record["input_hash_unchanged"]
print(record["selected_site"])
for name in ["F1", "F2"]:
    print(name, record["results"][name])
    print(record["mapping_checks"][name])
    assert record["results"][name]["uncovered_ground_fraction"] == 0
    assert record["mapping_checks"][name]["nonpositive_analytic_samples"] == 0
```

```python
for filename in ["view_search.png", "source_map.png", "composition_shortlist.png", "F1_F2_comparison.png", "physical_pair.png"]:
    display(Image(filename=str(OUTPUT / filename)))
```

## Interpret the comparison

F1's ground is physically plausible within the raw DEM model but much of its lower image magnifies a few 30 m triangles. F2 opens more geographic ground in the lower frame and balances near/middle/far coverage. Low/gentle statistics refer to native-face means, not actual accessibility or a mapped path.

F2 has an explicit 3D inverse when virtual depth is known; ordinary 2D terrain occlusion remains. No source tiles are relocated. Compare F2 as a new candidate before choosing line-generation controls; keep the previous P2/P3 mountain-mass images as alternatives.
