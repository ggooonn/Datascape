# Spring stroke hierarchy v0.2

Frozen F2 composition C (70/30). A is the body, B a structure accent, C fine directional grain and D a sparse final skeleton. Density, nominal width and dimensionless intensity are separate. This notebook does not search cameras or modify terrain/omission fields.

Install the project environment with `pip install -e ".[research,dev]"`; set `DATASCAPE_DATA_DIR` to the original TIFF directory.

```python
from pathlib import Path
import os, json
from IPython.display import display, Image
from datascape.rendering.hierarchy_study import run
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists())
data = Path(os.environ.get("DATASCAPE_DATA_DIR", root / "data/raw/seoul"))
output = Path(os.environ.get("DATASCAPE_HIERARCHY_OUTPUT", root / "outputs/spring_stroke_hierarchy_v02"))
config = root / "configs/scenes/seoul_spring_stroke_hierarchy_v02.json"
print(json.loads(config.read_text())["styles"])
```

## Four nested combinations

A and the archived vegetation context are identical in all four sets. C density variants share exact surviving paths, width and opacity. D is selected after omission, with a measured length ceiling, one-sided support and NMS spacing. Its taper lives in the exported vertex-width/intensity profile.

```python
record = run(data / "SeoulNorth_v02_MASTER_30m.tif", data / "SeoulNorth_v02_VEG_10m.tif", config, output, root)
display(Image(filename=str(output / "comparison_black.png")))
display(Image(filename=str(output / "comparison_density.png")))
display(Image(filename=str(output / "physical_comparison.png")))
```

## Density and final skeleton

Black preview: opacity 1 for every path. Density preview: the same geometry, with declared preview opacity; opacity is not physical pressure. Hierarchy colors are diagnostic only. Cliff and foothold rims remain DEM proxies, not verified rock edges.

```python
from datascape.strokes.vector import read_vectors
assert all(record["invariants"].values())
assert record["C_density"]["nested_subset"]
assert record["checks"]["terrain_core_sample_violations"] == 0
assert record["checks"]["all_footprints_empty_in_bottom_2mm"]
final = record["metrics"]["04_ABCD"]
assert final["D_terrain_length_fraction"] <= 0.06
assert final["D_all_length_fraction"] <= 0.06
print(json.dumps(final, indent=2))
print(record["D_selection"])
display(Image(filename=str(output / "C_density_comparison.png")))
display(Image(filename=str(output / "D_before_after_details.png")))
display(Image(filename=str(output / "final_hierarchy.png")))
```
