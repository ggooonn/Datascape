# Spring Pair Production Study v0.1

Extend native MASTER terrain toward the foreground while retaining P2/P3 depth anchors. Compare continuous 600 x 900 mm content and two 300 mm halves separated physically by 40 mm. This notebook generates terrain previews and mapped-space checks, not brush paths or G-code.

Install `python -m pip install -e ".[research,dev]"` from the repository, select that environment in VSCode, and set `DATASCAPE_DATA_DIR` to the original MASTER folder. Previous exploration outputs and original artworks stay unchanged.

```python
import os
from pathlib import Path
from IPython.display import Image, display
from datascape.rendering.spring_production import run

ROOT = next((p for p in [Path.cwd(), Path.cwd().parent] if (p / "pyproject.toml").is_file()), None)
if ROOT is None:
    raise FileNotFoundError("Run this notebook from the repository or notebooks directory.")
DATA_DIR = Path(os.environ.get("DATASCAPE_DATA_DIR", str(ROOT / "data/raw")))
MASTER = DATA_DIR / "SeoulNorth_v02_MASTER_30m.tif"
SOURCE_SCENE = ROOT / "configs/scenes/seoul_projection_composition_v01.json"
CONFIG = ROOT / "configs/scenes/seoul_spring_production_v01.json"
OUTPUT = Path(os.environ.get("DATASCAPE_SPRING_OUTPUT", str(ROOT / "outputs/spring_production_v01")))
if not MASTER.is_file():
    raise FileNotFoundError("Set DATASCAPE_DATA_DIR to the original MASTER folder.")
```

```python
report = run(MASTER, SOURCE_SCENE, CONFIG, OUTPUT)
assert report["input_hash_unchanged"] and report["raw_dem_preserved"]
assert report["spatial_checks_pass"], "Inspect source coverage, virtual depth and mapping diagnostics before proceeding."
for method in ["P2", "P3"]:
    print(method, report["results"][method])
    print("Mapping checks:", report["mapping_checks"][method])
    print("Visibility:", report["visibility_checks"][method])
```

```python
for filename in ["source_extension_map.png", "foreground_before_after.png", "spring_comparison.png", "physical_pair.png"]:
    display(Image(filename=str(OUTPUT / filename)))
```
