# A-density v2: mountain-relative breathing

F2-C and the selected foreground are locked. Only mountain A placement changes. B/C/D/V and foreground A are copied from the exact PR #11 baseline. Production alpha remains unchanged; opaque diagnostics are evaluation views, not material prescriptions.

```python
from pathlib import Path
import os, sys, json
from IPython.display import display, Image
here=Path.cwd().resolve()
root=next(p for p in [here,*here.parents] if (p/"src/datascape").is_dir())
sys.path.insert(0,str(root/"src"))
master=Path(os.environ.get("DATASCAPE_MASTER",root/"data/raw/SeoulNorth_v02_MASTER_30m.tif"))
veg=Path(os.environ.get("DATASCAPE_VEG",root/"data/raw/SeoulNorth_v02_VEG_10m.tif"))
out=Path(os.environ.get("DATASCAPE_OUTPUT",root/"docs/experiments/spring_a_density_v02"))
cp=root/"configs/scenes/seoul_spring_a_density_v02.json"
assert master.is_file() and veg.is_file(), "Set DATASCAPE_MASTER and DATASCAPE_VEG to the original TIFFs."
print(json.loads(cp.read_text())["study"])
```

```python
from datascape.rendering.mass_density_study import run
record=run(master,veg,cp,out,root)
```

```python
from datascape.strokes.vector import read_vectors
from datascape.strokes.density_rhythm import layer_digest
baseline=read_vectors(root/json.loads(cp.read_text())["baseline_vectors"])["strokes"]
updated=read_vectors(out/"v2_strokes.json.gz")["strokes"]
for layer in ["B","C","D","V"]:
    assert layer_digest(baseline,layer)==layer_digest(updated,layer)
assert record["invariants"]["retention_float64_unchanged"]
assert record["invariants"]["nonpositive_phase_derivative_pixels"]==0
print(json.dumps(record["invariants"],indent=2))
```

The analysis catchments are not named geographic mountains or newly observed objects. Local density is indexed by each catchment's relief, not global screen height. Original q-phase values supply boundary row budgets; local integration runs in source meters. The unchanged mist transfer handles omission after placement.

Diagnostic C uses opaque blue at 0.60 mm for visibility; production C remains its saved 0.16 mm nominal style and original alpha profile. A-only diagnostics normalize width to 0.28 mm and alpha to 1. All displayed centerlines come from saved or explicitly identified candidate vectors.
