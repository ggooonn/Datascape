# Stroke Grammar Refinement v0.2 — color-ready spring pair

Keep the selected F2-C scene, original DEM/NDVI, 70/30 foreground, camera/projection and existing whitespace fields fixed. Compare actual A/C path density, render-space terminal D and internal B. The limited object prototype is an independent optional experiment.

Install from the repository root: `python -m pip install -e ".[research,dev]"`. Set `DATASCAPE_DATA_DIR` to the directory containing both original TIFFs. No raw input is written, and no G-code or machine action is performed.

```python
from pathlib import Path
import os, json
from IPython.display import display, Image
from datascape.rendering.color_ready_study import run, styled_segments
from datascape.strokes.vector import read_vectors
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "configs/scenes/seoul_spring_color_ready_v02.json").is_file())
DATA = Path(os.environ.get("DATASCAPE_DATA_DIR", ROOT / "data/raw"))
OUT = Path(os.environ.get("DATASCAPE_COLOR_READY_OUTPUT", ROOT / "outputs/spring_color_ready_v02"))
CONFIG = ROOT / "configs/scenes/seoul_spring_color_ready_v02.json"
master = DATA / "SeoulNorth_v02_MASTER_30m.tif"
veg = DATA / "SeoulNorth_v02_VEG_10m.tif"
assert master.is_file() and veg.is_file(), "Set DATASCAPE_DATA_DIR to the original TIFF directory."
```

```python
record = run(master, veg, CONFIG, OUT, ROOT)
```

## Density is geometry; transparency is a display simulation

Current / ink-dominant / color-ready share the same B/D/V. New A paths come from actual half-depth source rows. C uses nested selections from the existing direction-field proposals. The same canonical paths and width profiles produce opaque black and transparency-aware previews. Alpha is not dried-ink opacity or physical brush pressure.

## D: visible terminal boundaries, then connected breathing

Skyline peak shoulders, front-side depth/source discontinuities and gated visible ridge/cliff candidates are geometric sources. A endpoints only contribute a small secondary score. The artist's red reference is not an input to path generation. Parent arclength controls the envelope continuously through the panel seam; straight paths retain interior profile controls after geometry simplification.

```python
display(Image(filename=str(ROOT / "docs/experiments/spring_color_ready_v02/reference_red_annotation.png")))
display(Image(filename=str(OUT / "D_comparison.png")))
display(Image(filename=str(OUT / "D_detail_comparison.png")))
display(Image(filename=str(OUT / "D_edge_diagnostic.png")))
record["D"]
```

## B: intrinsic structure before mist retreat

Candidate paths and salience come from unmasked DEM derivatives, slope breaks, ridge and valley structure. The existing continuous retention and A support are applied afterward. Adjustable retreat and an orientation penalty suppress paths parallel to strong mist transitions, rather than creating a permanent horizontal exclusion band.

```python
display(Image(filename=str(OUT / "B_comparison.png")))
display(Image(filename=str(OUT / "B_detail_comparison.png")))
record["B"]["boundary_relation"]
```

## Review the primary color-ready combination

Color-ready A/C + revised internal B + connected terminal D + current vegetation. No final color mixing is implemented. The 40 mm physical gap is inserted after the 600 × 900 mm content is split at x = 300 mm; content is not cropped to create the gap.

```python
display(Image(filename=str(OUT / "recommended_physical.png")))
display(Image(filename=str(OUT / "layer_diagnostics.png")))
record["metrics"]["recommended"], record["overlap"]["recommended"]
```

```python
drawing = read_vectors(OUT / "recommended.json.gz")["strokes"]
config = json.loads(CONFIG.read_text())
opaque = styled_segments(drawing, "black", config)
composite = styled_segments(drawing, "composite", config)
import numpy as np
assert len(opaque[0]) == len(composite[0])
for a, b in zip(opaque[0], composite[0]):
    np.testing.assert_array_equal(a, b)
np.testing.assert_array_equal(opaque[1], composite[1])
assert record["invariants"]["nominal_bottom_2mm_empty"]
assert record["invariants"]["terrain_zero_core_violations"] == 0
assert record["invariants"]["density_only_B_D_V_identical"]
record["invariants"]
```
