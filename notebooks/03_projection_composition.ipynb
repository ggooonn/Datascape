# Projection & Composition Study v0.1

Compare B-V3 spatial translations on continuous 600 x 900 mm content. The 40 mm physical gap is inserted afterward. Original terrain and artworks remain unchanged. No brush strokes, vegetation, color, or G-code are generated.

Install `python -m pip install -e ".[research,dev]"` from the repository and select that environment in VSCode. Set `DATASCAPE_DATA_DIR` to the folder containing MASTER. The CLI and this notebook use the same reusable modules. See `docs/experiments/projection_composition_v01/README.md` for projection definitions, reference sources and limits.

```python
import os
from pathlib import Path
from IPython.display import Image, display
from datascape.rendering.composition_study import run

REPO_ROOT = next((p for p in [Path.cwd(), Path.cwd().parent] if (p / "pyproject.toml").is_file()), None)
if REPO_ROOT is None:
    raise FileNotFoundError("Run from the repository or notebooks directory.")
DATA_DIR = Path(os.environ.get("DATASCAPE_DATA_DIR", str(REPO_ROOT / "data/raw")))
MASTER = DATA_DIR / "SeoulNorth_v02_MASTER_30m.tif"
CONFIG = REPO_ROOT / "configs/scenes/seoul_projection_composition_v01.json"
OUTPUT = Path(os.environ.get("DATASCAPE_PROJECTION_OUTPUT", str(REPO_ROOT / "outputs/projection_composition_v01")))
if not MASTER.is_file():
    raise FileNotFoundError("Set DATASCAPE_DATA_DIR to the folder containing the original MASTER GeoTIFF.")
```

```python
record = run(MASTER, CONFIG, OUTPUT)
assert record["raw_dem_preserved"] and record["input_hash_unchanged"]
print(record["input"])
print(record["projection"])
for method, result in record["results"].items():
    print(method, [(b["label"], round(b["visible_terrain_fraction"], 4), round(b["centroid_y_fraction"], 4)) for b in result["bands"]])
```
